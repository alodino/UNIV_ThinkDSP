## Create Square Signals

```python
from thinkdsp import SquareSignal

signal = SquareSignal(freq=1100)

wave = signal.make_wave(duration=0.5, framerate=10000)
```

## Plot Sampled Wave

```python
wave.segment(start=0, duration=0.005).plot()

plt.title(f'Square Wave')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid(True)
plt.show()
```

## Compute and Plot Wave Spectrum

```python
# Compute Spectrum
spectrum = wave.make_spectrum()

# Plot full spectrum
spectrum.plot()

plt.title(f'Spectrum')

plt.xlabel('Frequency (Hz)')
plt.ylabel('Amplitude')
plt.grid(True)

plt.show()
```
